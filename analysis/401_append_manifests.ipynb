# Append manifest files

This notebook appends two manifest files together to create a compiled manifest for future runs.

```python
import os
from pathlib import Path
from datetime import datetime

import dotenv
import pandas as pd

# Load environment variables
dotenv.load_dotenv()

# Set up output folder
OUTPUT_FOLDER = Path(os.environ["OUTPUT_FOLDER"])
MANIFEST_DIR = OUTPUT_FOLDER / "individual_runs" / "manifests"
```

```python
# Define manifest files to append
MANIFEST_FILE_1 = MANIFEST_DIR / "2026-05-19_09-30_manifest.csv"
MANIFEST_FILE_2 = MANIFEST_DIR / "2026-05-20_15-29_manifest.csv"
```

```python
# Load both manifest files
manifest_1 = pd.read_csv(MANIFEST_FILE_1)
manifest_2 = pd.read_csv(MANIFEST_FILE_2)

print(f"Manifest 1: {len(manifest_1)} rows")
print(f"Manifest 2: {len(manifest_2)} rows")
```

```python
# Append the manifest files
combined_manifest = pd.concat([manifest_1, manifest_2], ignore_index=True)

print(f"Combined manifest: {len(combined_manifest)} rows")
combined_manifest
```

```python
# Save the combined manifest with timestamp
timestamp = datetime.now().strftime("%Y-%m-%d_%H-%M")
output_file = MANIFEST_DIR / f"{timestamp}_manifest_combined.csv"

combined_manifest.to_csv(output_file, index=False)
print(f"Saved combined manifest to: {output_file}")
```
